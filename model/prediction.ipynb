### Prediction with ANN model

```python
#Required Libraries
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import numpy as np
```

```python
#ANN model
model=load_model('model.keras')

#Encoded file 
with open('le_gen.pkl','rb') as file:
    le_gen=pickle.load(file)

with open('oneHot_geo.pkl','rb') as file:
    oneHot_geo=pickle.load(file)

#scaled file
with open('scaler.pkl','rb') as file:
    scaler=pickle.load(file)
```

```python
#Setting up the input data
input_data={
'CreditScore':600,
'Gender':'Male',
 'Age':40, 
 'Tenure': 3,
'Balance':60000,
'NumOfProducts':2,
'HasCrCard':1,
'IsActiveMember':1,
'EstimatedSalary':50000,
'Geography':'France'
}
```

```python
#converting the data to dataframe
input_df=pd.DataFrame([input_data])
input_df
```

```python
# OneHot encoding 
geo_encoded=oneHot_geo.transform([[input_data['Geography']]]).toarray()
geo_encoded_df=pd.DataFrame(geo_encoded,columns=oneHot_geo.get_feature_names_out(['Geography']))
```

```python
# label encoding 
input_df['Gender']=le_gen.transform([input_df['Gender']])
```

```python
# Concatination of onehot encoded data
input_df=pd.concat([input_df.drop('Geography',axis=1),geo_encoded_df],axis=1)
```

```python
input_df
```

```python
# scale the input data
input_scaler=scaler.transform(input_df)
input_scaler
```

```python
prediction=model.predict(input_scaler)
prediction
```

```python
prediction_proba=prediction[0][0]
```

```python
prediction_proba
```

```python
if prediction_proba > 0.5:
    print('The customer is likely to churn.')
else:
    print('The customer is not likely to churn.')
```
